# Manual Confocal Prematch GUI

This lightweight notebook uses `social_imaging_scripts.analysis.manual_confocal_prematch_gui`
to launch the refactored manual prematch interface.


```python
from pathlib import Path
from social_imaging_scripts.metadata.config import load_project_config
from social_imaging_scripts.metadata.loader import load_animals
from social_imaging_scripts.analysis.manual_confocal_prematch_gui import run_manual_prematch_gui

cfg = load_project_config()
animals = list(load_animals(base_dir=Path.cwd().parent / 'metadata' / 'animals').animals)

# Optional: focus on a single session
TARGET_ANIMAL = 'L395_f11'
TARGET_CONF_SESSION = None  # e.g. 'L395_f11_confocal_roundn_2025-07-04'

stats = run_manual_prematch_gui(
    animals=animals,
    cfg=cfg,
    force=True,  # set True to redo even if saved
    target_animal_id=TARGET_ANIMAL,
    target_confocal_session_id=TARGET_CONF_SESSION,
)
stats
```
